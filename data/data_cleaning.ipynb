```python
import re

def extract_court_type(court_name):
    """
    Extracts the type of court (e.g., 'Εφ' for Εφετείο) from the abbreviated name.

    Args:
        court_name (str): The abbreviated court name (e.g., 'ΜονΕφΠειρ').

    Returns:
        str or None: The extracted court type abbreviation, or None if not found.
    """

    cleaned_name = re.sub(r'\s*\(.*?\)\s*', '', court_name)

    # Common Greek court type abbreviations and their full names
    court_types = {
        'Εφ': 'Εφετείο',
        'ΜΟΕ': 'Μικτό Ορκωτό Εφετείο',
        'ΣυμβΕφ': 'Συμβούλιο Εφετών',
        'Πρ': 'Πρωτοδικείο',
        'Ειρ': 'Ειρηνοδικείο',
        'ΣτΕ': 'Συμβούλιο της Επικράτειας',
        'ΣΤΕ': 'Συμβούλιο της Επικράτειας',
        'ΑναθΔικ': 'Αναθεωρητικό Δικαστήριο',
        'ΑΠ': 'Άρειος Πάγος',
        'ΑΠΔ': 'Αρχή Προστασίας Δεδομένων',
        'ΕλΣ': 'Ελεγκτικό Συνέδριο',
        'ΑΕΔ': 'Ανώτατο Ειδικό Δικαστήριο',
        'Διαιτ': 'Διαιτητικό Δικαστήριο',
        'Διατ': 'Διάταξη',
        'ΟΠΕΚΑ': 'Οργανισμός Προνοιακών Επιδομάτων και Κοινωνικής Αλληλεγγύης',
        'Ναυτ': 'Ναυτικό'
    }

    # Sort keys by length descending to match longer, more specific terms first
    sorted_keys = sorted(court_types.keys(), key=len, reverse=True)

    for key in sorted_keys:
        if key in cleaned_name:
            return court_types[key]

    return '-' # Return '-' if no known type is found
```

```python
df["Τύπος δικαστηρίου"] = df["Απόφαση δικαστηρίου"].apply(extract_court_type)
df.head()
```
